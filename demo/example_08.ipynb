将一颗色子掷 6000 次，统计每种点数出现的次数。

```python
import random

f1, f2, f3, f4, f5, f6 = 0, 0, 0, 0, 0, 0

for _ in range(6000):
    point = random.randrange(1, 7)
    if point == 1:
        f1 = f1 + 1
    elif point == 2:
        f2 = f2 + 1
    elif point == 3:
        f3 = f3 + 1
    elif point == 4:
        f4 = f4 + 1
    elif point == 5:
        f5 = f5 + 1
    else:
        f6 = f6 + 1

print(f1, f2, f3, f4, f5, f6)
```

在 Python 中，列表是由一系列元素按特定顺序构成的数据序列，这就意味着如果我们定义一个列表类型的变量，可以用它来保存多个数据。在 Python 中，可以使用[]字面量语法来定义列表，列表中的多个元素用逗号进行分隔，代码如下所示。

```python
items = [35, 12, 99, 68, 55, 35, 87]
items2 = ['Python', 'Java', 'Go', 'Kotlin']
items3 = [100, 12.3, 'Python', True]

items4 = list(range(1, 10))
print(items4)

items5 = list('hello')
print(items5) 
```

如果希望一次性访问列表中的多个元素，我们可以使用切片运算。切片运算是形如[start:end:stride]的运算符，其中start代表访问列表元素的起始位置，end代表访问列表元素的终止位置（终止位置的元素无法访问），而stride则代表了跨度，简单的说就是位置的增量，比如我们访问的第一个元素在start位置，那么第二个元素就在start + stride位置，当然start + stride要小于end。我们给上面的代码增加下面的语句，来使用切片运算符访问列表元素。

```python
items9 = ['Python', 'Java', 'JavaScript', 'CPython', 'CJava', 'abcd']
a = items9[1:4:1]
# range 参数的用法 [start, end, stride]
for i in items9[::2]:
    print(i, end=', ')
```
